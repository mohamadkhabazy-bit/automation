```python
import os

# Workaround for the common OpenMP duplicate-DLL crash on Anaconda/Windows
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

import faulthandler
import time

faulthandler.enable()

AUDIO = r"C:\Users\RAYANSTAR\Downloads\audio_2026-11-05_17-43-57.ogg"
CACHE = r"E:\whisper_models"


def step(msg):
    print(msg, flush=True)


step("1) import faster_whisper")
from faster_whisper import WhisperModel

step("2) load model medium on CPU (int8)")
t = time.time()
model = WhisperModel("medium", device="cpu", compute_type="int8", download_root=CACHE)
step(f"   loaded in {time.time() - t:.1f}s")

step("3) transcribe WITHOUT vad")
t = time.time()
segments, info = model.transcribe(AUDIO, language="fa", beam_size=5,
                                  vad_filter=False, temperature=0.0)
text = " ".join(s.text.strip() for s in segments)
step(f"   done in {time.time() - t:.1f}s")
step("TEXT: " + text)

step("4) transcribe WITH vad")
t = time.time()
segments, info = model.transcribe(AUDIO, language="fa", beam_size=5,
                                  vad_filter=True, temperature=0.0)
text = " ".join(s.text.strip() for s in segments)
step(f"   done in {time.time() - t:.1f}s")
step("TEXT: " + text)
```
