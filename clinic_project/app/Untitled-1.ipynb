```python

import os
import sys
import subprocess
import platform
import urllib.request
import urllib.error
import socket

def run_command(title, command):
    print("\n" + "=" * 70)
    print(title)
    print("=" * 70)

    try:
        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            timeout=20,
            shell=True
        )

        print(result.stdout)

        if result.stderr:
            print("STDERR:")
            print(result.stderr)

        print("Exit code:", result.returncode)

    except Exception as e:
        print("ERROR:", e)


def check_url(url):
    print(f"\nTesting: {url}")

    try:
        req = urllib.request.Request(
            url,
            headers={"User-Agent": "Mozilla/5.0"}
        )

        with urllib.request.urlopen(req, timeout=12) as response:
            print("SUCCESS - HTTP Status:", response.status)

    except Exception as e:
        print("FAILED:", repr(e))


def main():

    print("=" * 70)
    print("PYTHON / PIP / NETWORK DIAGNOSTICS")
    print("=" * 70)

    print("\nPython version:", sys.version)
    print("Python executable:", sys.executable)
    print("Operating system:", platform.platform())

    run_command(
        "[1] PIP VERSION",
        f'"{sys.executable}" -m pip --version'
    )

    run_command(
        "[2] PIP CONFIGURATION",
        f'"{sys.executable}" -m pip config list'
    )

    run_command(
        "[3] PROXY ENVIRONMENT VARIABLES",
        'set | findstr /i "proxy"'
        if os.name == "nt"
        else 'env | grep -i proxy'
    )

    run_command(
        "[4] DEFAULT DNS",
        "nslookup files.pythonhosted.org"
    )

    run_command(
        "[5] GOOGLE DNS",
        "nslookup files.pythonhosted.org 8.8.8.8"
    )

    run_command(
        "[6] CLOUDFLARE DNS",
        "nslookup files.pythonhosted.org 1.1.1.1"
    )

    run_command(
        "[7] PIP CACHE",
        f'"{sys.executable}" -m pip cache list'
    )

    print("\n" + "=" * 70)
    print("[8] PYPI CONNECTION")
    print("=" * 70)
    check_url("https://pypi.org/simple/pymysql/")

    print("\n" + "=" * 70)
    print("[9] PYTHON PACKAGE SERVER CONNECTION")
    print("=" * 70)
    check_url("https://files.pythonhosted.org/")

    print("\n" + "=" * 70)
    print("[10] GOOGLE DNS HTTPS CONNECTION")
    print("=" * 70)
    check_url("https://dns.google/")

    print("\n" + "=" * 70)
    print("[11] DNS RESOLUTION TEST")
    print("=" * 70)

    for host in [
        "pypi.org",
        "files.pythonhosted.org",
        "google.com"
    ]:
        try:
            ip = socket.gethostbyname(host)
            print(f"{host}: {ip}")
        except Exception as e:
            print(f"{host}: FAILED - {e}")

    print("\n" + "=" * 70)
    print("[12] PYMYSQL INSTALL TEST")
    print("=" * 70)

    run_command(
        "INSTALL PYMYSQL",
        f'"{sys.executable}" -m pip install pymysql --no-cache-dir -v'
    )

    print("\n" + "=" * 70)
    print("DIAGNOSTICS COMPLETED")
    print("=" * 70)


if __name__ == "__main__":
    main()
```
