```python
import time
import av
import numpy as np
from faster_whisper import WhisperModel


def load_audio_16k(path: str) -> np.ndarray:
    """Decode any audio file to mono float32 at 16 kHz."""
    resampler = av.audio.resampler.AudioResampler(format="s16", layout="mono", rate=16000)
    chunks = []

    def collect(result):
        if result is None:
            return
        frames = result if isinstance(result, list) else [result]
        for f in frames:
            chunks.append(f.to_ndarray().reshape(-1))

    with av.open(path, mode="r") as container:
        for frame in container.decode(audio=0):
            collect(resampler.resample(frame))
    collect(resampler.resample(None))  # flush

    return np.concatenate(chunks).astype(np.float32) / 32768.0


model = WhisperModel("small", device="cuda", compute_type="float16",
                     download_root=r"E:\whisper_models")

audio = load_audio_16k(r"C:\Users\RAYANSTAR\Downloads\audio_2026-11-05_17-43-57.ogg")
print("Audio seconds:", round(len(audio) / 16000, 1))

t = time.time()
segments, info = model.transcribe(
    audio, language="fa", beam_size=5, vad_filter=True, temperature=0.0,
)
text = " ".join(s.text.strip() for s in segments)
print("Time:", round(time.time() - t, 1), "seconds")
print("Text:", text)
```
